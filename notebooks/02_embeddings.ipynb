```python
embedding = model.encode(clean_df["text"].iloc[0])
```

```python
clean_df = pd.read_csv("../data/cleaned/healthcare_cleaned.csv")
```

```python
embedding.shape
```

```python
embeddings = model.encode(
    clean_df["text"].tolist(),
    show_progress_bar=True
)
```

```python
embeddings.shape
```

```python
import faiss
```

```python
import numpy as np
```

```python
embeddings = np.asarray(embeddings).astype("float32")
```

```python
faiss.normalize_L2(embeddings)
```

```python
index = faiss.IndexFlatIP(384)
```

```python
index.add(embeddings)
```

```python
index.ntotal
```

```python
query = "What are the symptoms of pink eye?"
```

```python
query_embedding = model.encode([query])
```

```python
query_embedding = np.asarray(query_embedding).astype("float32")
```

```python
faiss.normalize_L2(query_embedding)
```

```python
scores, indices = index.search(query_embedding, 3)
```

```python
indices
```

```python
scores
```

```python
for idx, score in zip(indices[0], scores[0]):
    print("Score:", score)
    print("Title:", clean_df.iloc[idx]["title"])
    print("-" * 80)
```

```python
import os
import faiss

os.makedirs("../backend/vector_store", exist_ok=True)

faiss.write_index(
    index,
    "../backend/vector_store/healthcare.index"
)

print("FAISS index saved successfully!")
```

```python
import pickle

with open("../backend/vector_store/documents.pkl", "wb") as f:
    pickle.dump(clean_df.to_dict("records"), f)

print("Documents saved successfully!")
```
